```python
def load_documents_with_metadata(data_dir, metadata_list):
    all_docs = []
    for i, file_path in enumerate(data_dir):
        loader = TextLoader(file_path, encoding="utf-8")
        docs = loader.load()
        docs[0].metadata = metadata_list[i]
        all_docs.extend(docs)
    return all_docs

def split_documents(docs, chunk_size=300, chunk_overlap=50):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
    )
    return text_splitter.split_documents(docs)
```

```python
docs = load_documents_with_metadata(DATA_DIR, DOCUMENT_METADATA)
splits = split_documents(docs)
```

```python
from dotenv import load_dotenv

load_dotenv()

from langchain_pinecone import PineconeVectorStore
from pinecone import Pinecone, ServerlessSpec

def data_ingestion_to_pinecone(splits, embedding_model, index_name="basiccare-guidelines"):
    # Initialize Pinecone
    pc = Pinecone()
    
    # Create a serverless index if it doesn't exist
    if not pc.has_index(index_name):
        pc.create_index(
            name=index_name,
            dimension=2048,
            metric="cosine",
            spec=ServerlessSpec(
                cloud='aws',
                region='us-east-1'
            )
        )
    docsearch = PineconeVectorStore.from_documents(
        documents=splits,
        index_name=index_name,
        embedding=embedding_model
    )
    
    print("Data is successfully ingested")
```

```python
# vectorstore
def vectordb(index_name):
    vector_store = PineconeVectorStore(
        index_name=index_name,
        embedding=embedding_model
    )
    return vector_store
```

```python
# retriever
def create_retriever(index_name, search_type="similarity", search_kwargs=None):
    vector_store = vectordb(index_name)
    retriever = vector_store.as_retriever(
        search_type=search_type,
        search_kwargs=search_kwargs
    )
    return retriever
```

```python
retriever = create_retriever(index_name="basiccare-guidelines", search_type="similarity", search_kwargs={"k": 4})
```

```python
# testing retriever
query = "At what temperature should someone seek professional help for a fever?"

docs = retriever.invoke(query)

for i, doc in enumerate(docs):
    print(f"\n--- RESULT {i+1} ---")
    print("DOC ID:", doc.metadata.get("doc_id"))
    print("VERSION:", doc.metadata.get("version"))
    print(doc.page_content)
```
